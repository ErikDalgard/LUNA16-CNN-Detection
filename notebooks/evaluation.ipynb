```python
import os
import glob
from tensorflow import keras
from evaluation_functions import plot_roc, plot_pr, plot_sensitivity_fp, ensemble_predict
```

```python
#Loading the test data 
X_test_2D = 1
X_test_3D = 1
X_test_ae = 1

y_test = 1
val_aucs = {} #read from CSV logger
```

```python
#Getting the paths of each models best weight and adding it to a family list
model_paths_2D = glob.glob("training_history/2D*/best_model_*.h5", recursive=True)
model_paths_3D = glob.glob("training_history/3D*/best_model_*.h5", recursive=True)
model_paths_AE = glob.glob("training_history/AE*/best_model*.h5", recursive=True)

families = {
    "2D CNN": {"paths": model_paths_2D, "X_test": X_test_2D},
    "3D CNN": {"paths": model_paths_3D, "X_test": X_test_3D},
    "2D AE": {"paths": model_paths_AE, "X_test": X_test_ae},}
```

```python
models = []

for family_name, family in families.items():
    print(f"Evaluationg {family_name}")
    predictions = {}
    for path in family["paths"]:
        model_name = path.split(os.sep)[-2]
        print(f"Loading {model_name}")
        model = keras.models.load_model(path)

        y_predict = model.predict(family["X_test"]).ravel()

        predictions[model_name] = y_predict

    y_ensemble, weights = ensemble_predict(model_paths=family["paths"], X_test=family["X_test"], val_aucs=val_aucs[family_name])

    plot_roc(y_test, predictions, family_name)
    plot_pr(y_test, predictions, family_name)
    plot_sensitivity_fp(y_test, predictions, family_name)
```
