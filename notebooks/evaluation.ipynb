```python
import os
import glob
from tensorflow import keras
from evaluation_functions import plot_roc, plot_pr, plot_sensitivity_fp
```

```python
X_test = 1
y_test = 1
```

```python
model_weights_paths = glob.glob("training_history/**/best_model_*.h5", recursive=True)
```

```python
for path in model_weights_paths:
    model_name = path.split(os.sep)[-2]
    print(f"Loading {model_name}")
    model = keras.models.load_model(path)

    y_predict = model.predict(X_test).ravel()

    plot_roc(y_test, y_predict, model_name)
    plot_pr(y_test, y_predict, model_name )
    plot_sensitivity_fp(y_test, y_predict, model_name)
```
